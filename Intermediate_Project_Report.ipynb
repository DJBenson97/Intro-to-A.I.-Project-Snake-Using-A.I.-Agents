### Project Overview

In this project, we are developing an A.I. agent capable of playing the game Snake. The agent navigates a grid based environment, collecting food while avoiding collisions with both its own body and the enviironment's walls. The snake agent grows one cell larger each time it makes contact with the randomly generated and placed food. The game's difficulty constantly increases as the snake devours more food, due to the fact that the ever increasing size of the snake makes it much more difficult to navigate the grid and seek food without ever colliding with itself or the walls.

We have implemented multiple A.I. agents using search algorithms and probabilistic reasoning. These algorithms include:
- Random (the agent moves around without strategy, selecting from available valid actions)
- Greedy Best-First Search
- Breadth-First Search (BFS)
- A* Search
- Several enhanced A* Search variants that incorporate Bayesian risk estimation and survival based logic.

### Current Progress

At this stage, the Snake game environment has been fully implemented and is functioning correctly. It successfully supports multiple A.I. agents, each capable of making decisions based on the current state of the environment.

The following agents have been successfully implemented and tested:
- Random Agent
- Greedy Best-First Search Agent
- Breadth-First Search Agent
- A* Search Agent
- A* with Bayesian Risk (Local)
- A* With Tail-Reachability (Survival)

Further, we have also implemented a simulation framework that allows us to run multiple games (currently 100) automatically and then collect performance data for each agent. This performance data is then used to construct a Bar Graph for better visualization and comparison for the user.

---

### Demonstration of Functionality

We have verified that the current system functions correctly by running visual simulations of each agent. These simulations show that:
- The Random agent moves without strategy, thus performing poorly.
- The Greedy agent moves directly towards food, but can sometimes be prone to trapping itself.
- BFS finds valid paths to food and performs better in structured environments.
- A* improves upon BFS by using heuristics for more efficient movement.
- The Bayesian and Tail-Safe agents demonstrate improved survival by avoiding risky or trapping situations.

All agents currently run without errors across multiple simulations, confirming that the implementation is stable. Further, the results gathered from running each simulation 100 times are in line with the behavior that you would expect from each of the various algorithms.

---

### Results So Far

We conducted multiple simulation runs for each agent and collected performance metrics. These metrics include minimum score, maximum score, and average score.

The results show clear differences in performance between the various algorithms. The Random agent performed very poorly, the worst by far as was expected, while the Greedy agent showed significant improvement but still struggled due to its short sighted decision making.

More advanced agents, such as the BFS and A* ones, achieved similar and substantially higher performance, indicating that both algorithms are effective at finding paths to a destination (food). However, neither of the agents fully acounts for long term survival.

The most advanced agents, particularly A* with tail-reachability logic, achieved the highest average scores and the most consistent performance, as shown by their higher minimum scores. This suggests that incorporating survival-based reasoning helps reduce failures and improves overall reliability.
