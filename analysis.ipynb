### Data Analysis

```python
agents = {
    "Random": RandomAgent(),
    "Greedy": GreedyAgent(),
    "BFS": BFSAgent(),
    "A*": AStarAgent(),

    # don't uncomment the line below unless you want the simulations to take a REALLY, really long time.
    # I ran the algorithms with the run count set to ONLY 10 and it still made it take
    # a full 2 minutes and 37 seconds, but when I commented it out the run time dropped
    # all the way down to just 3.6 seconds. I'll have to think about the Bayes implementation
    # a little bit more before submission, because this isn't tenable how it is. I added further
    # applications of Bayes though with AStarBayesLocalAgent and AStarTailSafeAgent.

    # "A* + Bayes (Full)": AStarBayesAgent(),
    
    "A* + Bayes (Local)": AStarBayesLocalAgent(),
    "A* + Tail Safe": AStarTailSafeAgent()
}
```

```python
results = {}

# lower the number if your pc is struggling to handle it but it should be fine at 100, it just takes a while
NUM_GAMES = 100 # ~3.6 seconds at 10 games, ~17.9 seconds at 50 games, ~35.5 seconds at 100 games (without "Bayes (Full)")

for name, agent in agents.items():
    scores = []

    for _ in range(NUM_GAMES):
        score = run_game(agent, render_game=False)
        scores.append(score)

    results[name] = scores
```

```python
summary = {}

for name, scores in results.items():
    avg_score = np.mean(scores)
    max_score = np.max(scores)
    min_score = np.min(scores)

    summary[name] = {
        "avg_score": avg_score,
        "max_score": max_score,
        "min_score": min_score
    }

    print(
        f"The {name} algorithm had a minimum score of {min_score}, "
        f"a maximum score of {max_score}, "
        f"and an average score of {avg_score:.2f}.\n"
    )
```

```python
names = list(summary.keys())
avg_scores = [summary[n]["avg_score"] for n in names]

plt.figure(figsize=(10,5))
bars = plt.bar(names, avg_scores)

plt.xticks(rotation=30)
plt.ylabel("Average Score")
plt.title("Agent Performance Comparison")

# numbers above bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f"{height:.2f}",
        ha='center',
        va='bottom'
    )

plt.show()
```
