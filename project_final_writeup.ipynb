## Overview of the Project

In this project, we developed an artificial intelligence system capable of playing the game Snake using a variety of algorithmic strategies. The goal was to design and compare multiple agents that operate within a grid-based environment, where the snake must collect food while avoiding collisions with both its own body and the boundaries of the grid.

The game environment maintains the current state, including the snake’s position, food location, and grid dimensions. Each agent is responsible for selecting actions based on this state. As the snake consumes food, it grows in length, increasing the complexity of navigation and making long-term survival more difficult.

The primary objective of the project was to evaluate how different decision-making strategies affect performance. We focused on comparing simple baseline approaches with more advanced search-based algorithms, analyzing how effectively each agent balances short-term goals (reaching food) with long-term survival.

For our project, we developed an artificial intelligence system capable of playing the game Snake using a variety of different algorithmic strategies. The goal was to design and then compare multiple agents that operate within a grid-based environment, where the snake must collect food while avoiding collisions with both its own body and the boundaries of the grid.

The game environment maintains the current state, including the snake's position, food location, the grid dimensions, and the boundaries surrounding the game arena. Each agent is responsible for selecting actions based on this state. The snake grows in length as it continues to consume food, which in turn increases the complexity of navigation and making survival more difficult as the game goes on, as it is Game Over if the snake collides with a wall or its own body. The walls remain static, so the true challenge of the game for the snake is to not collide with its ever increasing body.

## Algorithms Implemented

The following agents were implemented and tested in this project:

- **Random Agent**: Selects a valid move at random without any strategy. This serves as a kind of baseline for comparison.
- **Greedy Agent**: Moves toward the food using Manhattan distance, without considering long-term consequences or potential trapping positions.
- **Breadth-First Search (BFS)**: Explores all possible paths level by level in order to find the shortest path to the food.
- **A\***: Uses a combination of path cost and heuristic distance to efficiently compute an optimal path to the food.
- **Weighted A\***: Modifies A* by scaling the heuristic, making the agent more goal-directed and faster but potentially less optimal.
- **A\* with Tail-Reachability Safety**: Extends A* by ensuring that after reaching food, the snake can still reach its tail, improving its long-term survival while reducing self-trapping behavior.

Additionally, we experimented with an agent that incorporated a **global space heuristic**, which attempted to evaluate risk based on the amount of reachable space. However, due to its high computational cost and lack of performance improvement, it was not included in the final evaluation. It remains in the project, commented out, so we can discuss what went wrong and what we learned when we go through the interview grading.

## Code Functionality, Libraries, and Frameworks

The project is implemented in Python and is structured into three main components: the game environment, the AI agents, and the analysis framework.

The **game environment** (`snake_game.py`) handles all core mechanics of Snake, including movement, collision detection, food generation, and score tracking. It provides a function that runs a full simulation by repeatedly updating the game state based on the actions chosen by an agent.

The **agents** (`agents.py`) implement different strategies for selecting moves and determining the snake's path. Each agent follows the same interface by implementing a `get_action(state, valid_moves)` function. This modular design allows different algorithms to be easily swapped and compared under identical conditions.

The **analysis and visualization** (`snake_runner_and_analysis.ipynb`) is used to run simulations, collect performance data, and visualize results. The minimum, maximum, and average score for each agent is output after running 100 simulations each, and then the results are used to construct a bar graph.

The following libraries were used:
- **random**: for generating food locations and random agent behavior
- **heapq**: for implementing priority queues in search algorithms such as A*
- **collections.deque**: for efficient queue operations in BFS
- **numpy**: for grid representation and data handling
- **matplotlib.pyplot**: for rendering the game and visualizing results
- **time** and **IPython.display**: for controlling animation timing and updating visual output in simulations

No external frameworks were used beyond standard Python libraries and visualization tools.

## Testing Methodology and Results

To evaluate the effectiveness of each agent, we conducted a series of controlled experiments. Each agent was tested by running **100 independent simulations** of the Snake game. In each simulation, the agent played until it either collided or reached a maximum step limit.

For each agent, we recorded:
- Minimum score
- Maximum score
- Average score

This allowed us to measure both performance consistency and overall effectiveness.

The results from one set of experiments are as follows:

- The Random algorithm had a minimum score of 0, a maximum score of 4, and an average score of 0.60.
- The Greedy algorithm had a minimum score of 7, a maximum score of 35, and an average score of 24.32.
- The BFS algorithm had a minimum score of 21, a maximum score of 36, and an average score of 28.39.
- The A* algorithm had a minimum score of 22, a maximum score of 36, and an average score of 28.81.
- The A* (Weighted α=4) algorithm had a minimum score of 21, a maximum score of 36, and an average score of 28.88.
- The A* + Tail Safe algorithm had a minimum score of 22, a maximum score of 39, and an average score of 28.27.

These results demonstrate that more advanced algorithms significantly outperform simpler approaches. While A* and BFS provide strong pathfinding capabilities, the Tail Safe agent achieves improved survivability by incorporating long-term safety considerations.

## Example of Application

The application works by simulating a full game of Snake in which an AI agent controls the snake’s movement. At each step, the current game state is passed to the agent, which selects a valid move based on its strategy.

For example:
- A Greedy agent selects moves that minimize distance to the food.
- A BFS agent computes the shortest path through exhaustive search.
- An A* agent balances efficiency and optimality using heuristics.
- The Tail Safe agent evaluates whether reaching food will lead to a safe future state before committing to a path.

Visual simulations allow us to observe these behaviors in real time, making it clear how different strategies affect movement patterns, risk-taking, and overall performance. Each of the agents possess a code cell in 'snake_runner_and_analysis.ipynb' where the user can run a simulation and observe the gamestate as it progresses.

## Conclusion

In this project, we learned that effective AI behavior in Snake requires more than just simply finding the shortest path to food. While algorithms such as BFS and A* were able to significantly improve performance over other simpler approaches, they do not inherently account for the long-term survival of the snake. As the snake continues to grow, avoiding self-trapping becomes increasingly more important, which is where more advanced strategies become more obviously necessary.

On that note, the Tail Safe agent we implemented demonstrated the importance of incorporating future-state reasoning into decision-making. By ensuring that the snake could still reach its tail after pursuing food, it was able to avoid many of the failures seen in other agents, often becoming our strongest performing agent. This showed that balancing immediate rewards with long-term safety is critical in dynamic environments, as an inbalance between the two can easily lead to critical failures in decision making.

A major learning point from this project came from our attempts to implement Bayesian logic into our agents. Initially, we believed that modeling “risk” using probabilities would improve the agent’s decision making. However, our implementation did not actually follow true Bayesian principles, unfortunately. Instead of defining meaningful random variables, priors, and updating beliefs based on observed data, we approximated risk using heuristic measures such as the amount of reachable space. While this may resemble probabilistic reasoning at first glance, it is fundamentally different from Bayesian inference.

Because of this, the variables we used were effectively **uninformative in a Bayesian sense**. They did not represent true probabilities or learned distributions, and they were not updated based on new information. As a result, our “Bayesian” agents were simply heuristic based agents with additional computational overhead. This not only failed to improve performance, but in some cases actually made the agents significantly slower without any measurable benefit. Bayesian methods require carefully defined, informative variables and a proper probabilistic framework, and without these components, they can easily become inefficient heuristics.

This experience also showed that Snake may not be the best game environment for applying Bayesian reasoning in the way we initially intended. The game is largely deterministic with full state observability, which means that there is little inherent uncertainty to model. Without uncertainty, the advantages of Bayesian inference are limited, which made it difficult to apply the concept in a meaningful way.

Overall, this project demonstrated the strengths and limitations of different AI strategies, and while also making it clear how important it is to combine efficient search with thoughtful consideration of future consequences when designing intelligent agents. It also taught us more about how Bayesian logic functions, although the process of learning this was very frustrating. We hope to discuss this more at length during our interview grading, as we believe there to be much value in having an experienced instructor to bounce ideas off of for complex concepts such Bayesian logic.
